## Fonctions reprises de l'exercice 01

````python
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)
````

**Rappel de syntaxe** — le system prompt n'est pas un message de la liste, c'est un paramètre à part :

```python
client.messages.create(
    model=model,
    max_tokens=1000,
    system="Tu es ...",   # ← ici
    messages=messages,
)
```

````python
def chat(messages, system=None):

    params = {
        "model": model,
        "max_tokens":1000,
        "messages": messages,
    }
    if system:
        params["system"] = system
        
    message = client.messages.create(**params)
    for bloc in message.content:
        if bloc.type == "text":
            return bloc.text
````

````python
messages = []

while(True):
    text = input()
    if text == "quit":
      break
    add_user_message(messages, text)
    answer = chat(messages, system="Tu es un developeur senior python qui créer les fonctions les plus concises possible")
    print(answer)
    add_assistant_message(messages, answer)
````
